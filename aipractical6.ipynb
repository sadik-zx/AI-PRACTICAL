```python
# ================================================================
# PRACTICAL 6
# CONSTRAINT SATISFACTION PROBLEM (CSP)
# MAP COLORING USING BACKTRACKING AND CONSISTENCY CHECKING
# ================================================================

# ------------------------------------------------
# AIM
# ------------------------------------------------
# To solve a Constraint Satisfaction Problem (CSP)
# such as Map Coloring using Backtracking and
# Consistency Checking.
#
# CSP consists of:
# 1. Variables
# 2. Domains
# 3. Constraints
#
# In this practical:
# Variables  -> Regions A, B, C, D
# Domain     -> Red, Green, Blue
# Constraint -> Connected regions cannot have same color
# ------------------------------------------------


print("=" * 70)
print("PRACTICAL 6: CONSTRAINT SATISFACTION PROBLEM")
print("Map Coloring using Backtracking and Consistency Checking")
print("=" * 70)


# ================================================================
# STEP 1: DEFINE THE MAP
# ================================================================
#
# Map representation:
#
#        A -------- B
#        |        / |
#        |      /   |
#        |    /     |
#        C -------- D
#
# Connections:
# A -> B, C
# B -> A, C, D
# C -> A, B, D
# D -> B, C
#
# We represent the map using a Python dictionary.


map_data = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}


# ================================================================
# STEP 2: DEFINE THE DOMAIN
# ================================================================
#
# Domain means the possible values that can be
# assigned to every variable.
#
# Here, the possible colors are:
# Red, Green and Blue.


colors = ['Red', 'Green', 'Blue']


print("\nMAP CONNECTIONS")
print("-" * 30)

for region, neighbors in map_data.items():
    print(f"{region} -> {neighbors}")


print("\nAVAILABLE COLORS")
print("-" * 30)
print(colors)


# ================================================================
# STEP 3: CONSISTENCY CHECKING
# ================================================================
#
# Before assigning a color to a region, we check
# whether any of its already-colored neighbors
# have the same color.
#
# If a conflict exists:
#       return False
#
# If there is no conflict:
#       return True


def is_consistent(region, color, assignment):

    # Check every neighbor of the selected region
    for neighbor in map_data[region]:

        # We only check neighbors that have already
        # been assigned a color.
        if neighbor in assignment:

            # If the neighbor has the same color,
            # the constraint is violated.
            if assignment[neighbor] == color:
                return False

    # No conflict found
    return True


# ================================================================
# STEP 4: DEMONSTRATE CONSISTENCY CHECKING
# ================================================================

print("\n" + "=" * 70)
print("CONSISTENCY CHECKING DEMONSTRATION")
print("=" * 70)


test_assignment = {
    'A': 'Red'
}


print("\nCurrent assignment:")
print(test_assignment)


print("\nCan we assign B = Red?")
print(
    "Result:",
    is_consistent('B', 'Red', test_assignment)
)


print("\nCan we assign B = Green?")
print(
    "Result:",
    is_consistent('B', 'Green', test_assignment)
)


# ================================================================
# STEP 5: BACKTRACKING SOLVER
# ================================================================
#
# Backtracking works using the following strategy:
#
# 1. Select an unassigned region.
# 2. Try a color.
# 3. Check whether the color is consistent.
# 4. If consistent, assign the color.
# 5. Recursively solve the remaining problem.
# 6. If the choice causes failure:
#       remove the assignment.
#       try another color.
#
# This process of undoing a previous decision
# is called BACKTRACKING.


def solve_map(assignment):

    # ------------------------------------------------------------
    # BASE CASE
    # ------------------------------------------------------------
    #
    # If all regions have been assigned a color,
    # the CSP has been successfully solved.

    if len(assignment) == len(map_data):

        return assignment.copy()


    # ------------------------------------------------------------
    # SELECT AN UNASSIGNED REGION
    # ------------------------------------------------------------

    region = next(
        region for region in map_data
        if region not in assignment
    )


    # ------------------------------------------------------------
    # TRY EVERY COLOR
    # ------------------------------------------------------------

    for color in colors:

        # --------------------------------------------------------
        # CONSISTENCY CHECK
        # --------------------------------------------------------

        if is_consistent(region, color, assignment):

            # ----------------------------------------------------
            # ASSIGN THE COLOR
            # ----------------------------------------------------

            assignment[region] = color


            # ----------------------------------------------------
            # RECURSIVE CALL
            # ----------------------------------------------------
            #
            # Try to solve the remaining regions.

            result = solve_map(assignment)


            # ----------------------------------------------------
            # IF SOLUTION IS FOUND
            # ----------------------------------------------------

            if result is not None:

                return result


            # ----------------------------------------------------
            # BACKTRACK
            # ----------------------------------------------------
            #
            # If the current assignment does not lead
            # to a solution, remove it and try another color.

            del assignment[region]


    # ------------------------------------------------------------
    # FAILURE
    # ------------------------------------------------------------
    #
    # If no color works for this region,
    # return None.

    return None


# ================================================================
# STEP 6: SOLVE THE MAP
# ================================================================

print("\n" + "=" * 70)
print("SOLVING MAP COLORING USING BACKTRACKING")
print("=" * 70)


# Start with an empty assignment

solution = solve_map({})


# ================================================================
# STEP 7: DISPLAY THE SOLUTION
# ================================================================

print("\nFINAL SOLUTION")
print("-" * 30)


if solution is not None:

    for region, color in solution.items():

        print(f"Region {region} = {color}")

else:

    print("No valid solution exists.")


# ================================================================
# STEP 8: VERIFY THE SOLUTION
# ================================================================
#
# We now independently check every connection.
#
# The rule is:
#
# Connected regions must have different colors.


def verify_solution(solution):

    for region, neighbors in map_data.items():

        for neighbor in neighbors:

            if solution[region] == solution[neighbor]:

                return False

    return True


# ================================================================
# STEP 9: DETAILED VERIFICATION
# ================================================================

print("\n" + "=" * 70)
print("DETAILED CONSTRAINT VERIFICATION")
print("=" * 70)


checked_pairs = set()

all_valid = True


for region, neighbors in map_data.items():

    for neighbor in neighbors:

        # Avoid checking the same pair twice.
        # Example:
        # A-B and B-A represent the same connection.

        pair = tuple(sorted([region, neighbor]))


        if pair not in checked_pairs:

            checked_pairs.add(pair)


            color1 = solution[region]
            color2 = solution[neighbor]


            if color1 != color2:

                print(
                    f"PASS: {region} ({color1}) "
                    f"!= {neighbor} ({color2})"
                )

            else:

                print(
                    f"FAIL: {region} ({color1}) "
                    f"== {neighbor} ({color2})"
                )

                all_valid = False


print("\nFINAL VERIFICATION:")

if all_valid:

    print("SUCCESS - All constraints are satisfied.")

else:

    print("FAILED - Some constraints are violated.")


# ================================================================
# STEP 10: BACKTRACKING TRACE
# ================================================================
#
# The following function is used only for demonstration.
#
# It shows:
#
# TRY
# CHECK
# ASSIGN
# CONFLICT
# BACKTRACK
# SUCCESS
#
# This helps students understand what happens
# internally during recursion.


def solve_map_with_trace(assignment, depth=0):

    # Indentation represents recursion depth.

    indent = "    " * depth


    # ------------------------------------------------------------
    # BASE CASE
    # ------------------------------------------------------------

    if len(assignment) == len(map_data):

        print(
            indent +
            "SUCCESS -> Complete solution found:",
            assignment
        )

        return assignment.copy()


    # ------------------------------------------------------------
    # SELECT REGION
    # ------------------------------------------------------------

    region = next(
        region for region in map_data
        if region not in assignment
    )


    print(
        indent +
        f"Selecting region: {region}"
    )


    # ------------------------------------------------------------
    # TRY COLORS
    # ------------------------------------------------------------

    for color in colors:

        print(
            indent +
            f"Trying {region} = {color}"
        )


        # --------------------------------------------------------
        # CHECK CONSISTENCY
        # --------------------------------------------------------

        if is_consistent(region, color, assignment):

            print(
                indent +
                f"    PASS -> Assign {region} = {color}"
            )


            # Assign color

            assignment[region] = color


            # ----------------------------------------------------
            # RECURSIVE CALL
            # ----------------------------------------------------

            result = solve_map_with_trace(
                assignment,
                depth + 1
            )


            # ----------------------------------------------------
            # IF SOLUTION FOUND
            # ----------------------------------------------------

            if result is not None:

                return result


            # ----------------------------------------------------
            # BACKTRACK
            # ----------------------------------------------------

            print(
                indent +
                f"    BACKTRACK -> Remove "
                f"{region} = {color}"
            )


            del assignment[region]


        else:

            print(
                indent +
                f"    CONFLICT -> "
                f"{region} cannot be {color}"
            )


    # No valid color found

    print(
        indent +
        f"FAILURE -> No valid color for {region}"
    )


    return None


# ================================================================
# STEP 11: RUN BACKTRACKING TRACE
# ================================================================

print("\n" + "=" * 70)
print("BACKTRACKING STEP-BY-STEP TRACE")
print("=" * 70)


trace_solution = solve_map_with_trace({})


# ================================================================
# STEP 12: FINAL RESULT
# ================================================================

print("\n" + "=" * 70)
print("FINAL RESULT")
print("=" * 70)


print("\nMap Coloring Solution:")

for region in map_data:

    print(
        f"Region {region} -> "
        f"{trace_solution[region]}"
    )


print("\nSolution Validity:")

if verify_solution(trace_solution):

    print("TRUE - The solution satisfies all constraints.")

else:

    print("FALSE - The solution violates constraints.")


# ================================================================
# PRACTICAL CONCLUSION
# ================================================================

print("\n" + "=" * 70)
print("PRACTICAL CONCLUSION")
print("=" * 70)

print("""
The Map Coloring problem was successfully solved as a
Constraint Satisfaction Problem (CSP).

Variables:
A, B, C and D

Domain:
Red, Green and Blue

Constraint:
Connected regions must have different colors.

Technique Used:
Backtracking with Consistency Checking.

Backtracking allows the algorithm to undo a previous
assignment when that assignment leads to failure and
try another possible value.

Therefore, CSP problems such as Map Coloring can be
solved systematically using backtracking and
constraint checking.
""")

print("=" * 70)
print("PRACTICAL 6 COMPLETED SUCCESSFULLY")
print("=" * 70)
```
